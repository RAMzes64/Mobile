```
class Person ()
{
    public string Name{get; init;}
    public int Age{get; set;}
    public override string ToString() => $"{Name} \t {Age}";

    public void Deconstruct(out string name, out int age, out string text)
    {
        name = Name;
        age = Age;
        text = "sdfsd";
    }
}

var p1 = new Person() {Name = "Alex", Age = 89};
var p2 = new Person() {Name = "Alex", Age = 89};

(string name, int age, _) = p1;
System.Console.WriteLine(name + "\t" + age);

(_, int age1, string text) = p2;
System.Console.WriteLine(age1 + "\t" + text);
```

```
class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public override string ToString() => $"{Name} \t {Age}";

    public virtual void Print()
    {
        System.Console.WriteLine(ToString());
    }   
    public Person(string name, int age)
    {
        Name = name;
        Age = age;
    }
}

class Employee : Person
{
    public string Company { get; set; }

    public Employee(string name, int age, string company) : base(name, age)
    {
        Company = company;
    }
}

var e = new Employee("Alex", 23, "Revolut");
e.Print();
```

```
class Code(string text)
{
    public string Text {get;} = text;

    public int Length
    {
        get => Text.Length;
    }

    public string this[Index index]
    {
        get => new string(Text[index], 1);
    }

    public IEnumerator<char> GetEnumerator(){
        foreach (var item in Text)
        {
            yield return item;
        }
    }
}

var code = new Code("Привет!");

System.Console.WriteLine(code[^2]);

for (int i = 1; i <= code.Length; i++)
{
    System.Console.WriteLine(code[^i]);
}
```
