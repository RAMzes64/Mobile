# Генератор функций

```
IEnumerator GetCities()
{
    yield return "Калуга";
    yield return "Воронеж";
    yield return "Тула";
}

var city17 = GetCities();
city17.MoveNext();
System.Console.WriteLine(city17.Current);
```

```
IEnumerable<string> GetCities()
{
    yield return "Калуга";
    yield return "Воронеж";
    yield return "Тула";    
}
foreach (var i in GetCities())
{
    System.Console.WriteLine(i);
}
```

```
IEnumerable<int> GetNumber(int start, int end)
{
    for (var i = start; i < end; i++)
    {
        yield return i;
    }
}

foreach (var item in GetNumber(1, 5))
{
    System.Console.WriteLine(item); 
}
```
