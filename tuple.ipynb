# Кортеж

```
var i = 1;
var j = (1, 2);

System.Console.WriteLine(j.Item1);
```

```
(int a, int b) tuple = (2, 3);

System.Console.WriteLine(tuple.a);
```

```
var tuple1 = (a: 1, b:2);
System.Console.WriteLine(tuple1.b);
```

```
var (a, b) = (2,3);
System.Console.WriteLine(a + "\t" + b);
```

```
(int a, int b) getTuple() => (2,3);

var s = getTuple();
System.Console.WriteLine(s);
```

```
(int a, int b) getNewTuple(int v, (int j, int k) jk) => (v + jk.j + jk.k, 0);

System.Console.WriteLine(getNewTuple(3, (1, 5)));
```
